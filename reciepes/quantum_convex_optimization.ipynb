# Quantum Spin-Chain State & Convex Optimization Recipe

This recipe combines 3 `algebrax` tools to analyze quantum spin states and convex optimization:

1. **Fenchel-Legendre Convex Conjugate** (`algebrax.transforms.legendre_fenchel`):
   Computes dual Legendre-Fenchel conjugate values $f^*(s) = \sup_x \{ s \cdot x - f(x) \}$.
2. **Block Diagonal Hamiltonians & Trace** (`algebrax.matrix.core.block_diag` & `trace`):
   Constructs direct sum composite Hamiltonians $H = H_1 \oplus H_2$ and calculates matrix trace invariants $\text{Tr}(H)$.
3. **Probabilistic NFA Superposition Automata** (`algebrax.automata.simulate_nfa`):
   Simulates NFA state transitions across probabilistic quantum channels.

```python
from algebrax.automata import simulate_nfa
from algebrax.matrix.core import block_diag, trace
from algebrax.transforms import legendre_fenchel
```

## 1. Fenchel-Legendre Convex Conjugate (legendre_fenchel)

We compute dual conjugate values $f^*(s)$ for $f(x) = \frac{1}{2} x^2$.

```python
convex_signal = {x: 0.5 * (x**2) for x in range(-5, 6)}
slopes = [-2.0, -1.0, 0.0, 1.0, 2.0]

print('Legendre-Fenchel Dual Conjugate Values f*(s):')
for s in slopes:
    print(f'  Slope s = {s:+4.1f} -> f*(s) = {legendre_fenchel(convex_signal, slope=s):6.2f}')
```

## 2. Multi-Qubit Block Diagonal Hamiltonian & Trace

We compute composite Hamiltonian $H = H_1 \oplus H_2$ and trace $\text{Tr}(H)$.

```python
h1 = {0: {0: 1.0, 1: 0.5}, 1: {0: 0.5, 1: -1.0}}
h2 = {0: {0: 2.0, 1: 0.1}, 1: {0: 0.1, 1: -2.0}}
composite = block_diag([h1, h2])
tr_h = trace(composite)

print('Composite Block Diagonal Matrix H = H1 (+) H2:', composite)
print(f'Hamiltonian Trace Tr(H): {tr_h:.2f}')
```

## 3. Probabilistic Quantum Decay Automaton (simulate_nfa)

We simulate NFA state transitions over probabilistic quantum decay channels.

```python
quantum_nfa = {0: {'pulse': {0: 0.5, 1: 0.5}}, 1: {'pulse': {1: 0.7, 2: 0.3}}, 2: {'pulse': {2: 1.0}}}
start_dist = {0: 1.0}
pulse_seq = ['pulse', 'pulse', 'pulse']

final_dist = simulate_nfa(start_dist, pulse_seq, quantum_nfa)
print('Final Quantum State Distribution:', final_dist)
```
