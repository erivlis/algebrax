# Schwarzschild Black Hole Spacetime & Gravitational Lensing Recipe

This recipe combines 5 `algebrax` tools to simulate Black Hole Physics and Gravitational Lensing:

1. **Schwarzschild Spacetime Metric Tensor** (`algebrax.tensor.einsum`):
   Contracts Schwarzschild spacetime metric tensors $g^{\mu \alpha} g_{\alpha \nu} = \delta^\mu_\nu$.
2. **Gravitational Redshift Z-Transform** (`algebrax.transforms.z_transform`):
   Evaluates complex frequency modulation $X(z)$ of light escaping from near the event horizon $r_s$.
3. **Gravitational Lensing & Spatial Curvature** (`algebrax.analysis.gradient` & `forman_ricci_curvature`):
   Computes gravitational deflection angles $\Delta \phi = \frac{4GM}{c^2 b}$ and photon sphere spatial curvatures.
4. **Bekenstein-Hawking Black Hole Entropy** (`algebrax.probability.entropy` & `kl_divergence`):
   Audits quantum information scrambling into Hawking radiation.

```python
import math

from algebrax.analysis import forman_ricci_curvature, gradient
from algebrax.probability import entropy, kl_divergence
from algebrax.tensor import einsum
from algebrax.transforms import z_transform
from algebrax.trie import AlgebraicTrie
```

## 1. Schwarzschild Spacetime Metric Tensor (einsum)

We construct Schwarzschild metric tensor $g_{\mu \nu}$ at $r = 2.0 r_s$ and verify contraction identities.

```python
r_s = 29.5  # km (Schwarzschild radius for 10 Solar Masses)
r_eval = 2.0 * r_s
f_r = 1.0 - (r_s / r_eval)

g_metric = AlgebraicTrie()
g_metric[(0, 0)] = -f_r
g_metric[(1, 1)] = 1.0 / f_r
g_metric[(2, 2)] = r_eval**2
g_metric[(3, 3)] = r_eval**2

g_inv = AlgebraicTrie()
g_inv[(0, 0)] = -1.0 / f_r
g_inv[(1, 1)] = f_r
g_inv[(2, 2)] = 1.0 / (r_eval**2)
g_inv[(3, 3)] = 1.0 / (r_eval**2)

identity = einsum('ma,an->mn', g_inv, g_metric)
print('Schwarzschild Metric Contraction Identity:', identity)
```

## 2. Gravitational Redshift Z-Transform (z_transform)

We evaluate frequency spectral shift $X(z)$ near the event horizon.

```python
z_red = (1.0 / math.sqrt(f_r)) - 1.0
h_signal = {0: 1.0, 1: 0.8, 2: 0.6, 3: 0.4, 4: 0.2}
z_complex = (1.0 + z_red) * (math.cos(math.pi / 4) + 1j * math.sin(math.pi / 4))

h_z = z_transform(h_signal, z=z_complex)
print(f'Redshifted Z-Transform H(z): {h_z:.4f}')
```

## 3. Gravitational Lensing Ray Deflection & Curvature

We evaluate Einstein light deflection angles and photon sphere spatial curvatures.

```python
b_impact = 3.0 * r_s
deflection_rad = 2.0 * r_s / b_impact
print(f'Gravitational Deflection Angle: {deflection_rad:.4f} rad ({math.degrees(deflection_rad):.2f} deg)')

spacetime_graph = {1: {2: 1.5}, 2: {1: 1.5, 3: 2.0}, 3: {2: 2.0, 4: 3.0}, 4: {3: 3.0}}
ricci_k = forman_ricci_curvature(spacetime_graph)
print('Forman-Ricci Curvature near Photon Sphere:', ricci_k)
```

## 4. Bekenstein-Hawking Black Hole Entropy

We calculate event horizon surface area and Hawking radiation information scrambling.

```python
p_infall = {0: 0.70, 1: 0.20, 2: 0.10}
q_hawking = {0: 0.34, 1: 0.33, 2: 0.33}

s_in = entropy(p_infall)
s_hawk = entropy(q_hawking)
kl_scramble = kl_divergence(p_infall, q_hawking)

print(f'Infilling Entropy S_in:          {s_in:.4f} bits')
print(f'Hawking Thermal Entropy S_hawk:  {s_hawk:.4f} bits')
print(f'Information Scrambling KL:       {kl_scramble:.4f} bits')
```
