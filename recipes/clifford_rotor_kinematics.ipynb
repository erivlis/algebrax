# Clifford Geometric Algebra $Cl(3,0)$ & 3D Rotor Rotations using `algebrax`

**Mathematical Foundations & Theory:**
1. **Geometric Product ($AB = A \\cdot B + A \\wedge B$)**: Combines inner and outer products over Clifford blade keys.
2. **Rotor Sandwiching ($v' = R v R^\\dagger$)**: Computes 3D spatial rotations $R = \\exp(-\\theta/2 \\mathbf{B})$ without gimbal lock.

```python
import math

from algebrax.clifford import CliffordSemiring, rotor_rotation

# 3D Spatial Vector v = 3 e1 + 4 e2
v = {(1,): 3.0, (2,): 4.0}
v_rot = rotor_rotation(v, bivector=(1, 2), angle_rad=math.pi / 2.0, p=3, q=0, r=0)
print(f"Rotated Vector v': e1={v_rot.get((1,), 0.0):.2f}, e2={v_rot.get((2,), 0.0):.2f}")
```
