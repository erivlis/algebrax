# Categorical Morphisms & Kleisli Monadic Composition using `algebrax`

**Mathematical Foundations & Theory:**
1. **Kleisli Category**: Morphisms $f: A \\to T(B)$ compose via Kleisli matrix multiplication ($g \\circ_T f$).
2. **Monadic Semiring Enrichment**: Swapping semirings defines probabilistic, cost-metric, and boolean reachability compositions.

```python
import algebrax as ax

# Probabilistic Kleisli Composition
f = {'A': {'B': 0.8, 'C': 0.2}}
g = {'B': {'D': 0.9}, 'C': {'D': 0.5}}
res = ax.category.kleisli_compose(f, g, semiring=ax.semiring.ViterbiSemiring())
print("Probabilistic Max Path A -> D:", res['A']['D'])
```
