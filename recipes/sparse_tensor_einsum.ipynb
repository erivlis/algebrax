# Sparse Tensor Einstein Summation & Multimodal Data Fusion Recipe

This recipe demonstrates tensor linear algebra using `algebrax.tensor` and `AlgebraicTrie`:

1. **Rank-3 Tensor Contraction** (`algebrax.tensor.einsum`):
   Contracts rank-3 tensors $C_{i, l} = \sum_{j, k} A_{i, j, k} \cdot B_{j, k, l}$ for multimodal data fusion.
2. **Tropical Min-Plus Tensor Optimization** (`algebrax.semiring.TropicalSemiring`):
   Executes `einsum` over `TropicalSemiring` for min-plus tensor network routing.
3. **Outer Product & Tensordot** (`algebrax.tensor.outer_product` & `tensordot`):
   Computes tensor outer products $C = A \otimes B$ and axis contractions.
4. **Nested Dictionary Converters** (`algebrax.tensor.flatten_tensor` & `unflatten_tensor`):
   Converts bidirectionally between nested dicts and tuple-indexed `AlgebraicTrie` mappings.

```python
import algebrax as ax
```

## 1. Rank-3 Tensor Contraction via einsum

We contract rank-3 multimodal tensors over shared indices $(j, k)$.

```python

tensor_a = ax.trie.AlgebraicTrie(semiring=ax.semiring.StandardSemiring)
tensor_a[('User_1', 'Movie_A', 'Home')] = 4.5
tensor_a[('User_1', 'Movie_B', 'Work')] = 2.0

tensor_b = ax.trie.AlgebraicTrie(semiring=ax.semiring.StandardSemiring)
tensor_b[('Movie_A', 'Home', 'Sci-Fi')] = 0.9
tensor_b[('Movie_B', 'Work', 'Comedy')] = 0.8

fused = ax.tensor.einsum('ijk,jkl->il', tensor_a, tensor_b)
print('Contracted Tensor C(i, l):', fused)
```

## 2. Tropical Min-Plus Tensor Contraction (TropicalSemiring)

We perform min-plus tensor contraction using `TropicalSemiring`.

```python
import algebrax as ax

trop_a = ax.trie.AlgebraicTrie(semiring=ax.semiring.TropicalSemiring)
trop_a[('N1', 'N2', 'Route_A')] = 10.0
trop_b = ax.trie.AlgebraicTrie(semiring=ax.semiring.TropicalSemiring)
trop_b[('N2', 'Route_A', 'Dest_X')] = 15.0

trop_fused = ax.tensor.einsum('ijk,jkl->il', trop_a, trop_b, semiring=ax.semiring.TropicalSemiring())
print('Tropical Min-Plus Latency Tensor:', trop_fused)
```

## 3. Outer Product & Tensordot (outer_product & tensordot)

We compute outer products and axis contractions.

```python
import algebrax as ax

v1 = {(0,): 2.0, (1,): 3.0}
v2 = {(0,): 4.0, (1,): 5.0}

out = ax.tensor.outer_product(v1, v2)
dot_res = ax.tensor.dot(v1, v2, axes=1)
print('Outer Tensor Product:', out)
print('Tensordot Result:    ', dot_res)
```

## 4. Nested Dictionary Tensor Converters (flatten_tensor & unflatten_tensor)

We perform roundtrip conversions between nested dicts and flat tensors.

```python
import algebrax as ax

nested = {'U1': {'M1': 4.5}, 'U2': {'M2': 3.5}}
flat = ax.tensor.flatten_tensor(nested)
reconstructed = ax.tensor.unflatten_tensor(flat)

print('Flattened Tensor:', flat)
print('Reconstructed:   ', reconstructed)
```
