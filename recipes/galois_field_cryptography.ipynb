# Galois Finite Field $\\text{GF}(2^8)$ & AES Cryptographic Arithmetic using `algebrax`

**Mathematical Foundations & Theory:**
1. **Galois Field $\\text{GF}(p^m)$**: Polynomial quotient arithmetic modulo an irreducible polynomial $P(x)$.
2. **AES MixColumns**: Matrix multiplication over $\\text{GF}(2^8)$ with $P(x) = x^8 + x^4 + x^3 + x + 1$.

```python
from algebrax.galois import GaloisFieldSemiring, gf_matrix_mul

gf = GaloisFieldSemiring(p=2, irreduc_poly=(1, 1, 0, 1, 1, 0, 0, 0, 1))
a = {4: 1}
b = {4: 1}
res_poly = gf.mul(a, b)
print("a * b mod P(x) in GF(2^8):", res_poly)
```
