# Algebraic Knot Theory & Topological Invariants Recipe

This recipe combines 3 `algebrax` tools to evaluate Algebraic Knot Theory and Skein Modules:

1. **Skein Module Connected Sums** (`algebrax.semiring.KnotSemiring`):
   Models formal linear combinations of knot topologies over the connected sum monoid $(\#)$.
2. **Braid Group Crossing Permutations & Signatures** (`algebrax.group.compose` & `signature`):
   Evaluates Artin braid strand crossings $B_n$ and crossing parity orientation signatures $\pm 1$.
3. **Jones Polynomial Invariant Arithmetic** (`algebrax.semiring.MonoidAlgebraSemiring`):
   Computes composite knot invariant multiplication $V(K_1 \# K_2) = V(K_1) \cdot V(K_2)$ over Laurent polynomial rings.

```python
import algebrax as ax
```

## 1. Skein Module Formal Sums & Connected Sums (KnotSemiring)

We multiply formal knot linear combinations using `KnotSemiring`.

```python

knot_semiring = ax.semiring.KnotSemiring(ax.semiring.StandardSemiring[float]())
knot_a = {'3_1': 2.0, 'U': 3.0}
knot_b = {'4_1': 1.0, 'U': 4.0}

composite = knot_semiring.mul(knot_a, knot_b)
print('Connected Sum Product A (#) B:', composite)
```

## 2. Braid Group Strand Crossings (compose & signature)

We compose Braid group generators $\sigma_1, \sigma_2$ and evaluate parity signatures.

```python
import algebrax as ax

s1 = {0: 1, 1: 0, 2: 2, 3: 3}
s2 = {0: 0, 1: 2, 2: 1, 3: 3}
w = ax.group.compose(ax.group.compose(s1, s2), s1)

print('Composed Braid Word w:', w)
print(f'Braid Parity Signature: {ax.group.signature(w):+d}')
```

## 3. Laurent Jones Polynomial Arithmetic (MonoidAlgebraSemiring)

We compute Jones polynomial multiplication $V(3_1 \# 4_1) = V(3_1) \cdot V(4_1)$.

```python
import algebrax as ax

poly_semiring = ax.semiring.MonoidAlgebraSemiring(ax.semiring.StandardSemiring[float](), zero_key=0)
v_trefoil = {-4: -1.0, -3: 1.0, -1: 1.0}
v_figure8 = {-2: 1.0, -1: -1.0, 0: 1.0, 1: -1.0, 2: 1.0}

v_comp = poly_semiring.mul(v_trefoil, v_figure8)
print('Composite Knot Jones Polynomial V(3_1 # 4_1):', v_comp)
```
