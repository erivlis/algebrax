# Control Theory, State-Space Systems & Dynamical Stability Recipe

This recipe combines 3 `algebrax` tools to evaluate Control Theory and Discrete State-Space Systems:

1. **Discrete State Transition Matrix Powers** (`algebrax.matrix.core.power`):
   Computes multi-step state propagation $x[k] = A^k x[0]$ over autonomous transition matrix $A$.
2. **Impulse Response Z-Transform** (`algebrax.transforms.z_transform`):
   Maps discrete impulse response sequences $h[n]$ into Z-domain transfer functions $H(z) = \sum h[n] z^{-n}$.
3. **Characteristic Stability Determinants** (`algebrax.matrix.academic.determinant`):
   Evaluates characteristic matrix determinants $\det(I - A)$ for asymptotic stability audits.

```python
import cmath

import algebrax as ax
```

## 1. Discrete State Transition Dynamics (A^k)

We compute multi-step state dynamics $A^k$ for a discrete 2x2 state-space system.

```python

a_matrix = {0: {0: 0.9, 1: 0.2}, 1: {0: -0.1, 1: 0.8}}
x0 = {0: 10.0, 1: 0.0}

for k in [1, 2, 5, 10]:
    a_k = ax.matrix.power(a_matrix, k)
    x1_k = a_k[0].get(0, 0.0) * x0[0] + a_k[0].get(1, 0.0) * x0[1]
    print(f'Step k={k:2d}: Position x1 = {x1_k:6.3f}')
```

## 2. Impulse Response Z-Transform Transfer Function (z_transform)

We evaluate Z-transform transfer function $H(z)$ at $z = 0.8 + 0.6j$.

```python
import algebrax as ax

h_n = {0: 1.0, 1: 0.5, 2: 0.25, 3: 0.125, 4: 0.0625}
z_val = 0.8 + 0.6j
h_z = ax.transforms.z_transform(h_n, z=z_val)
print(f'Transfer Function H(z): {h_z:.4f}')
```

## 3. Characteristic Matrix Stability Audit (determinant)

We evaluate determinant $\det(I - A)$ to audit system stability.

```python
import algebrax as ax

char_matrix = {0: {0: 0.1, 1: -0.2}, 1: {0: 0.1, 1: 0.2}}
det_val = ax.matrix.academic.determinant(char_matrix)
print(f'Characteristic Determinant det(I - A): {det_val:.4f}')
```
