# Topological Homology & Betti Barcode Invariants using `algebrax`

**Mathematical Foundations & Theory:**
1. **Simplicial Boundary Operator ($D_k$)**: Maps $k$-simplices to $(k-1)$-simplices via alternating boundary face sums.
2. **Homological Nilpotency**: $D_{k-1} \\circ D_k = \\mathbf{0}$ verified via sparse matrix multiplication.
3. **Betti Numbers ($\\beta_k$)**: Computes topological hole counts ($\\beta_0$ = components, $\\beta_1$ = 1D loops, $\\beta_2$ = 2D voids).

```python
from algebrax.homology import SimplicialComplex

# 1D Hollow Ring Topology (S^1)
ring_edges = [(0, 1), (1, 2), (2, 3), (0, 3)]
sc_ring = SimplicialComplex(ring_edges)
betti_ring = sc_ring.betti_numbers(max_k=1)
print(f"1D Ring Betti Numbers: beta_0 = {betti_ring[0]}, beta_1 = {betti_ring[1]}")

# 3D Solid Tetrahedron Topology
sc_tet = SimplicialComplex([(0, 1, 2, 3)])
betti_tet = sc_tet.betti_numbers(max_k=2)
print(f"Solid Tetrahedron Betti Numbers: beta_0 = {betti_tet[0]}, beta_1 = {betti_tet[1]}, beta_2 = {betti_tet[2]}")
```
